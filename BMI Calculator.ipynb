# BMI Calculator

```python
weight = float(input('Enter your weight in Kgs: '))
height = (float(input('Enter your height in cm: ')))/10

BMI = ((weight/(height**2)))*100

print('Your BMI is: ', BMI)

if BMI>0:
    if(BMI<=18.5):
        print("You're underweight!")
    elif(BMI<=24.9):
        print("You're have normal weight!")
    elif(BMI<=29.9):
        print("You're overweight!")
    elif(BMI<=34.9):
        print("You're obese!")
    elif(BMI<=39.9):
        print("You're severely obese!")
    else:
        print("You're morbidly obese!")
else:
    print("Enter valid inputs")
```
